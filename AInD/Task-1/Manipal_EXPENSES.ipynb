```python
outgo = []
in_come = []
credit = []
```

```python
"""This is the main code."""
while True:
  x = input("Enter an amount spent")
  if x == " ":
    break
  else:
   outgo.append(int(x))
while True:
  y = input("Enter the amount credited by others which you paid")
  if y == " ":
    break
  else:
   in_come.append(int(y))
while True:
 z = input("Enter the amount actually credited")
 if z == " ":
  break
 else:
  credit.append(int(z))
expenditure = sum(outgo) - sum(in_come)
credited_amount = sum(credit)
balance = (credited_amount) - (expenditure)

print(f"EXPENDITURE = {expenditure}\nTOTAL CREDITED AMOUNT = {credited_amount}\nBALANCE = {balance}")
```
